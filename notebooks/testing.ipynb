```python
import sys
import os

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))
from src.tensor_contraction import TensorNetwork, Node
import numpy as np
```

```python
node0 = Node(out_dims=[1, 1, 2])  # Node 0 connected to Node 1 (twice) and Node 2
node1 = Node(out_dims=[0, 0, 2])  # Node 1 connected to Node 0 (twice) and Node 2
node2 = Node(out_dims=[0, 1, 3])  # Node 2 connected to Node 0, Node 1, and Node 3
node3 = Node(out_dims=[2])        # Node 3 connected to Node 2

# Create the tensor network
tn = TensorNetwork(nodes=[node0, node1, node2, node3])

# Visualize the network
tn.draw_network(title="Example Tensor Network")
```

```python
cost = tn.merge_nodes(0, 1)
print(f"Contraction cost: {cost}")

# Visualize after merging
tn.draw_network(title="After Merging Nodes 0 and 1")
```
